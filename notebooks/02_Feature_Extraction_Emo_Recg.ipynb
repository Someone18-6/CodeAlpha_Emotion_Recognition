```python
import numpy as np
import pandas as pd
import librosa
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
emotion_mapping = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

data = []

for file_path in audio_files:
    filename = os.path.basename(file_path)
    parts = filename.split("-")

    emotion_code = parts[2]
    actor = parts[6].split(".")[0]

    data.append({
        "file_path": file_path,
        "emotion_code": emotion_code,
        "emotion": emotion_mapping[emotion_code],
        "actor": int(actor)
    })

df = pd.DataFrame(data)

print("Dataset shape:", df.shape)
df.head()
```

Extracting features from one audio file first

MFCC (40): captures characteristics of the voice and its sound quality.



Chroma (12): captures pitch-class energy information.



Mel spectrogram (128): captures how sound energy is distributed across Mel-frequency bands.

```python
def extract_features(file_path):
    audio, sample_rate = librosa.load(
        file_path,
        sr=22050,
        mono=True
    )

    mfcc = librosa.feature.mfcc(
        y=audio, sr=sample_rate, n_mfcc=40
    )

    chroma = librosa.feature.chroma_stft(
        y=audio, sr=sample_rate
    )

    mel = librosa.feature.melspectrogram(
        y=audio, sr=sample_rate, n_mels=128
    )

    features = np.concatenate([
        np.mean(mfcc, axis=1),
        np.mean(chroma, axis=1),
        np.mean(mel, axis=1)
    ])

    return features
```

```python
sample_features = extract_features(audio_files[0])

print("Feature vector shape:", sample_features.shape)
print("Total features:", len(sample_features))
print("First 10 features:", sample_features[:10])
```

```python
import time

features_list = []
labels = []

start_time = time.time()

for i, row in enumerate(df.itertuples(index=False), start=1):
    try:
        features = extract_features(row.file_path)

        features_list.append(features)
        labels.append(row.emotion)

    except Exception as e:
        print(f"Error processing {row.file_path}: {e}")

    if i % 100 == 0:
        print(f"Processed {i}/{len(df)} audio files")

end_time = time.time()

print("\nFeature extraction completed!")
print("Successfully processed:", len(features_list))
print("Total labels:", len(labels))
print("Time taken:", round(end_time - start_time, 2), "seconds")
```

```python
X = np.array(features_list)
y = np.array(labels)

print("Feature matrix shape:", X.shape)
print("Label array shape:", y.shape)
print("Unique emotions:", np.unique(y))
print("Missing values:", np.isnan(X).sum())
```

```python
import librosa
import numpy as np

def extract_improved_features(file_path):
    audio, sr = librosa.load(file_path, sr=22050, mono=True)

    # 1. MFCC features
    mfcc = librosa.feature.mfcc(
        y=audio, sr=sr, n_mfcc=40
    )

    # 2. Chroma features
    chroma = librosa.feature.chroma_stft(
        y=audio, sr=sr, n_chroma=12
    )

    # 3. Mel-spectrogram features
    mel = librosa.feature.melspectrogram(
        y=audio, sr=sr, n_mels=128
    )

    # Calculate mean and standard deviation
    features = np.concatenate([
        np.mean(mfcc, axis=1),
        np.std(mfcc, axis=1),

        np.mean(chroma, axis=1),
        np.std(chroma, axis=1),

        np.mean(mel, axis=1),
        np.std(mel, axis=1)
    ])

    return features

print("Improved feature extraction function created!")
```

```python
import time
import numpy as np

improved_features_list = []
improved_labels = []

start_time = time.time()

for i, row in enumerate(df.itertuples(index=False), start=1):
    try:
        features = extract_improved_features(row.file_path)

        improved_features_list.append(features)
        improved_labels.append(row.emotion)

    except Exception as e:
        print(f"Error processing {row.file_path}: {e}")

    if i % 100 == 0:
        print(f"Processed {i}/{len(df)} audio files")

X_improved = np.array(improved_features_list)
y_improved = np.array(improved_labels)

print("\nFeature extraction complete!")
print("Features shape:", X_improved.shape)
print("Labels shape:", y_improved.shape)
print(f"Time taken: {time.time() - start_time:.2f} seconds")
```
